```python
import pandas as pd

# بارگذاری داده‌ها
file_path = 'df_with_c2s.csv.gz'

try:
    # بارگذاری داده‌ها از فایل فشرده
    df = pd.read_csv(file_path, compression='gzip')

    # نمایش ۵ خط اول برای اطمینان
    from IPython.display import display
    display(df.head())

    # استخراج جملات سلولی
    cell_sentences = df['cell_sentence'].dropna()  # جملات سلولی از ستون 'cell_sentence'

    # تعداد واژه‌ها (ژن‌ها) در هر جمله سلولی
    gene_counts = cell_sentences.apply(lambda sentence: len(sentence.split()))

    # نمایش تعداد ژن‌ها (واژه‌ها) در هر جمله سلولی
    print("\nتعداد ژن‌ها (واژه‌ها) در هر جمله سلولی:")
    print(gene_counts.head())

except FileNotFoundError:
    print(f"فایل {file_path} پیدا نشد. لطفاً مسیر فایل را چک کنید.")
except Exception as e:
    print(f"خطایی رخ داد: {e}")
```

```python
# تابع برای شمارش تعداد ژن‌ها در هر جمله (با فرض اینکه هر کلمه یک ژن است)
def count_genes(sentence):
    # فرض می‌کنیم که هر کلمه یک ژن است
    genes = sentence.split()  # تقسیم جمله به کلمات
    return len(genes)

# اعمال تابع شمارش روی جملات مختلف
before_gene_count = df['cell_sentence'].apply(count_genes)
after_d1_gene_count = df['c2s_treated_d1_sentence'].apply(count_genes)
after_d2_gene_count = df['c2s_treated_d2_sentence'].apply(count_genes)
after_d12_gene_count = df['c2s_treated_d12_direct'].apply(count_genes)

# نمایش نتایج
print("تعداد ژن‌ها قبل از درمان:")
print(before_gene_count.head())

print("\nتعداد ژن‌ها پس از درمان با داروی 1:")
print(after_d1_gene_count.head())

print("\nتعداد ژن‌ها پس از درمان با داروی 2:")
print(after_d2_gene_count.head())

print("\nتعداد ژن‌ها پس از درمان با داروی 12:")
print(after_d12_gene_count.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# انتخاب سلول مورد نظر (سطر اول)
cell_id = 0

# استخراج جملات ژنی
before = df.loc[cell_id, 'cell_sentence'].split()
d1 = df.loc[cell_id, 'c2s_treated_d1_sentence'].split()
d2 = df.loc[cell_id, 'c2s_treated_d2_sentence'].split()
d12 = df.loc[cell_id, 'c2s_treated_d12_direct'].split()

# پیدا کردن ژن‌های مشترک در همه مراحل
common_genes = set(before) & set(d1) & set(d2) & set(d12)

# آماده‌سازی داده برای رسم
stages = ['Before', 'D1', 'D2', 'D12']
gene_lists = [before, d1, d2, d12]

plt.figure(figsize=(12, 8))

for gene in common_genes:
    ranks = []
    for gene_list in gene_lists:
        if gene in gene_list:
            ranks.append(gene_list.index(gene) + 1)  # رتبه از 1 شروع می‌شود
        else:
            ranks.append(np.nan)
    
    plt.plot(stages, ranks, marker='o', alpha=0.3, linewidth=1)

plt.xlabel('Treatment Stage', fontsize=14, fontweight='bold')
plt.ylabel('Gene Expression Rank', fontsize=14, fontweight='bold')
plt.title(f'Gene Expression Rank Changes for Cell {cell_id}', fontsize=16, fontweight='bold')
plt.gca().invert_yaxis()  # رتبه 1 بالای نمودار باشد
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"تعداد ژن‌های مشترک: {len(common_genes)}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# انتخاب سلول
cell_id = 0

# استخراج جملات ژنی
before = df.loc[cell_id, 'cell_sentence'].split()
d1 = df.loc[cell_id, 'c2s_treated_d1_sentence'].split()
d2 = df.loc[cell_id, 'c2s_treated_d2_sentence'].split()
d12 = df.loc[cell_id, 'c2s_treated_d12_direct'].split()

# پیدا کردن ژن‌های مشترک
common_genes = set(before) & set(d1) & set(d2) & set(d12)

# محاسبه تغییر رتبه
gene_changes = {}
stages = ['Before', 'D1', 'D2', 'D12']
gene_lists = [before, d1, d2, d12]

for gene in common_genes:
    ranks = []
    for gene_list in gene_lists:
        ranks.append(gene_list.index(gene) + 1)
    
    # محاسبه تغییر کل (Before → D12)
    rank_change = abs(ranks[-1] - ranks[0])
    gene_changes[gene] = {'ranks': ranks, 'change': rank_change}

# انتخاب Top 20 ژن با بیشترین تغییر
top_genes = sorted(gene_changes.items(), key=lambda x: x[1]['change'], reverse=True)[:20]

# رسم نمودار
plt.figure(figsize=(14, 8))

for gene, data in top_genes:
    plt.plot(stages, data['ranks'], marker='o', linewidth=2, label=gene, alpha=0.7)

plt.xlabel('Treatment Stage', fontsize=14, fontweight='bold')
plt.ylabel('Gene Expression Rank', fontsize=14, fontweight='bold')
plt.title(f'Top 20 Genes with Largest Rank Changes (Cell {cell_id})', fontsize=16, fontweight='bold')
plt.gca().invert_yaxis()  # رتبه 1 بالای نمودار
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# انتخاب سلول
cell_id = 0

# استخراج جملات ژنی
before = df.loc[cell_id, 'cell_sentence'].split()
d1 = df.loc[cell_id, 'c2s_treated_d1_sentence'].split()
d2 = df.loc[cell_id, 'c2s_treated_d2_sentence'].split()
d12 = df.loc[cell_id, 'c2s_treated_d12_direct'].split()

# ژن‌های مشترک
common_genes = set(before) & set(d1) & set(d2) & set(d12)

# ایجاد ماتریس رتبه
rank_matrix = []
gene_list = []

for gene in common_genes:
    ranks = [
        before.index(gene) + 1,
        d1.index(gene) + 1,
        d2.index(gene) + 1,
        d12.index(gene) + 1
    ]
    rank_matrix.append(ranks)
    gene_list.append(gene)

# تبدیل به DataFrame
rank_df = pd.DataFrame(rank_matrix, columns=['Before', 'D1', 'D2', 'D12'], index=gene_list)

# محاسبه تغییر کل
rank_df['Total_Change'] = abs(rank_df['D12'] - rank_df['Before'])

# مرتب‌سازی بر اساس بیشترین تغییر
rank_df_sorted = rank_df.sort_values('Total_Change', ascending=False)

# انتخاب Top 30 ژن
top_30 = rank_df_sorted.head(30)

# رسم Heatmap
plt.figure(figsize=(10, 14))
sns.heatmap(
    top_30[['Before', 'D1', 'D2', 'D12']], 
    cmap='RdYlGn_r', 
    annot=True, 
    fmt='d',
    cbar_kws={'label': 'Expression Rank'},
    linewidths=0.5
)
plt.title(f'Top 30 Genes - Expression Rank Changes (Cell {cell_id})', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Treatment Stage', fontsize=12, fontweight='bold')
plt.ylabel('Genes', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

cell_id = 0

# استخراج جملات ژنی
before = df.loc[cell_id, 'cell_sentence'].split()
d1 = df.loc[cell_id, 'c2s_treated_d1_sentence'].split()
d2 = df.loc[cell_id, 'c2s_treated_d2_sentence'].split()
d12 = df.loc[cell_id, 'c2s_treated_d12_direct'].split()

# ژن‌های مشترک
common_genes = set(before) & set(d1) & set(d2) & set(d12)

# آماده‌سازی داده برای Boxplot
data_for_plot = []

for gene in common_genes:
    data_for_plot.append({'Stage': 'Before', 'Rank': before.index(gene) + 1})
    data_for_plot.append({'Stage': 'D1', 'Rank': d1.index(gene) + 1})
    data_for_plot.append({'Stage': 'D2', 'Rank': d2.index(gene) + 1})
    data_for_plot.append({'Stage': 'D12', 'Rank': d12.index(gene) + 1})

plot_df = pd.DataFrame(data_for_plot)

# رسم Boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(data=plot_df, x='Stage', y='Rank', palette='Set2', order=['Before', 'D1', 'D2', 'D12'])
plt.title(f'Gene Expression Rank Distribution (Cell {cell_id})', fontsize=14, fontweight='bold')
plt.xlabel('Treatment Stage', fontsize=12, fontweight='bold')
plt.ylabel('Expression Rank', fontsize=12, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.violinplot(
    data=plot_df, 
    x='Stage', 
    y='Rank', 
    palette='muted',
    order=['Before', 'D1', 'D2', 'D12'],
    inner='quartile'
)
plt.title(f'Gene Expression Rank Distribution - Violin Plot (Cell {cell_id})', fontsize=14, fontweight='bold')
plt.xlabel('Treatment Stage', fontsize=12, fontweight='bold')
plt.ylabel('Expression Rank', fontsize=12, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

cell_id = 0

# استخراج لیست ژن‌ها
before = df.loc[cell_id, 'cell_sentence'].split()
d1 = df.loc[cell_id, 'c2s_treated_d1_sentence'].split()
d2 = df.loc[cell_id, 'c2s_treated_d2_sentence'].split()
d12 = df.loc[cell_id, 'c2s_treated_d12_direct'].split()

# ژن‌های مشترک بین همه مراحل
common_genes = set(before) & set(d1) & set(d2) & set(d12)

stages = ['Before', 'D1', 'D2', 'D12']
gene_lists = [before, d1, d2, d12]

# ساخت DataFrame طولانی (Long format)
data_for_plot = []

for gene in common_genes:
    for stage, gene_list in zip(stages, gene_lists):
        data_for_plot.append({
            'Gene': gene,
            'Stage': stage,
            'Rank': gene_list.index(gene) + 1
        })

plot_df = pd.DataFrame(data_for_plot)

# رسم Cat Plot (Boxplot)
g = sns.catplot(
    data=plot_df,
    x='Stage',
    y='Rank',
    kind='box',
    order=stages,
    height=6,
    aspect=1.4
)

g.fig.suptitle(
    f'Gene Expression Rank Distribution Across Treatment Stages (Cell {cell_id})',
    y=1.05,
    fontsize=14,
    fontweight='bold'
)

g.set_axis_labels('Treatment Stage', 'Expression Rank')
g.ax.invert_yaxis()  # Rank 1 در بالا
plt.tight_layout()
plt.show()
```
